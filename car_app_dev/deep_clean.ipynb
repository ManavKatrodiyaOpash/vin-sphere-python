```python
summary = (
    df.groupby(['make', 'model', 'trim', 'noOfPassengers', 'cylinders', 'weightInKg'])
      .agg(
          unique_body_types=('bodyType', 'nunique'),
          body_types=('bodyType', lambda x: list(sorted(set(x))))
      )
      .reset_index()
)

# Filter for conflicts (Make sure there is no comma at the end of this line!)
summary = summary[summary['unique_body_types'] > 1]

# Display the DataFrame
summary
```
