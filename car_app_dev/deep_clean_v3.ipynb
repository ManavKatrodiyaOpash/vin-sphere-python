```python
df_clean = df.copy()

df_clean["model_original"] = df_clean["model"]

df_clean["model"] = (
    df_clean["model"]
    .astype(str)
    .str.upper()
    .str.strip()
)
```

```python
import re
import numpy as np
def basic_model_clean(x):
    if pd.isna(x):
        return np.nan

    x = str(x).upper().strip()

    # normalize separators
    x = re.sub(r'[-_/]+', ' ', x)

    # remove repeated spaces
    x = re.sub(r'\s+', ' ', x)

    return x.strip()
```

```python
df_clean["model"] = df_clean["model"].apply(basic_model_clean)
```

```python
import re

arabic_models = (
    df_clean[
        df_clean["model"].fillna("").apply(
            lambda x: bool(re.search(r'[\u0600-\u06FF]', str(x)))
        )
    ]
    [["make","model"]]
    .drop_duplicates()
    .sort_values(["make","model"])
)

arabic_models
```
