```python
df = pd.read_csv("../Data/data_final_price/final_clean.csv")
```

```python
df_price = pd.read_csv("../Data/data_final_price/df.csv")
```

```python
match_cols = [
    "chassisNumber",
    "year",
    "make",
    "model"
]

# Convert all match columns to string
for col in match_cols:
    df[col] = df[col].astype(str).str.strip().str.upper()
    df_price[col] = df_price[col].astype(str).str.strip().str.upper()

# Remove duplicate vehicle definitions from df_price
price_lookup = (
    df_price[match_cols + ["price"]]
    .drop_duplicates(subset=match_cols, keep="first")
)

# Merge price
df = df.merge(
    price_lookup,
    on=match_cols,
    how="left"
)

print("Matched:", df["price"].notna().sum())
print("Missing:", df["price"].isna().sum())
```
