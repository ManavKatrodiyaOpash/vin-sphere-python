## weightInKg

```python
df["weightInKg"]
```

```python
df["weightInKg"].isnull().sum()
```

```python
df[df["weightInKg"] > 10000]
```

```python
df["bodyType"].unique()
```

```python
# 1. Sedans / Hatchbacks / Coupes shouldn't exceed ~3,500 kg (or ~7,500 lbs)
passenger_cars = ['SEDAN', 'HATCHBACK', 'COUPE', 'CONVERTIBLE', 'STATION WAGON']
df.loc[df['bodyType'].isin(passenger_cars) & (df['weightInKg'] > 4000), 'weightInKg'] = np.nan

# 2. SUVs / MPVs / Vans shouldn't exceed ~4,500 kg (or ~10,000 lbs)
large_personal = ['SUV', 'MPV', 'VAN', 'PANEL VAN', 'PICKUP', 'PICKUP DOUBLE CAB', 'PICKUP SINGLE CAB', 'PICK UP EXT CAB']
df.loc[df['bodyType'].isin(large_personal) & (df['weightInKg'] > 5000), 'weightInKg'] = np.nan

# 3. Heavy commercial vehicles can easily be over 10k. Set a realistic roof for them (e.g., 40,000 kg empty)
heavy_vehicles = ['BUS', 'MINI BUS', 'TRUCK', 'TANKER', 'HEAVY EQUIPMENT', 'FORKLIFT', 'SINGLE CABIN LONG CHASSIS']
df.loc[df['bodyType'].isin(heavy_vehicles) & (df['weightInKg'] > 45000), 'weightInKg'] = np.nan
```

```python
df["weightInKg"].isnull().sum()
```

```python
df[df["weightInKg"] < 200]
```

```python
# 1. Sedans / Hatchbacks / Coupes (Fixes entries less than 600 kg)
passenger_cars = ['SEDAN', 'HATCHBACK', 'COUPE', 'CONVERTIBLE', 'STATION WAGON']
df.loc[df['bodyType'].isin(passenger_cars) & ((df['weightInKg'] > 4000) | (df['weightInKg'] < 600)), 'weightInKg'] = np.nan

# 2. SUVs / MPVs / Vans / Pickups (Fixes entries less than 1,000 kg)
large_personal = ['SUV', 'MPV', 'VAN', 'PANEL VAN', 'PICKUP', 'PICKUP DOUBLE CAB', 'PICKUP SINGLE CAB', 'PICK UP EXT CAB']
df.loc[df['bodyType'].isin(large_personal) & ((df['weightInKg'] > 5000) | (df['weightInKg'] < 1000)), 'weightInKg'] = np.nan

# 3. Heavy Commercial Vehicles (Fixes entries less than 2,500 kg)
heavy_vehicles = ['BUS', 'MINI BUS', 'TRUCK', 'TANKER', 'HEAVY EQUIPMENT', 'FORKLIFT', 'SINGLE CABIN LONG CHASSIS']
df.loc[df['bodyType'].isin(heavy_vehicles) & ((df['weightInKg'] > 45000) | (df['weightInKg'] < 2500)), 'weightInKg'] = np.nan

# 4. Fill all these new NaN values with the proper median for each body type
df['weightInKg'] = df.groupby('bodyType')['weightInKg'].transform(lambda x: x.fillna(x.median()))
```

```python
df[df["weightInKg"] < 100]
```

```python
# Add MOTORCYCLE to your cleaning pipeline
# Min threshold: 50 kg (accounts for small electric/petrol mopeds)
# Max threshold: 500 kg (accounts for heavy cruisers/touring bikes; anything above is an error)
df.loc[(df['bodyType'] == 'MOTORCYCLE') & ((df['weightInKg'] > 500) | (df['weightInKg'] < 50)), 'weightInKg'] = np.nan

# Re-run your group median fill to cleanly fix the newly flagged motorcycle NaNs
df['weightInKg'] = df.groupby('bodyType')['weightInKg'].transform(lambda x: x.fillna(x.median()))
```
