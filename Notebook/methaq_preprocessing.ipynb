```python
data = data.loc[data['make'].map(data['make'].value_counts()).sort_values(ascending=False).index]
data
```

```python
# 1. Clean data: Using already cleaned 'data' (where chassisNumber is not None or "None")
print(f"Original shape: {df_methaq.shape}")
print(f"Cleaned shape: {data.shape}")
print(f"Dropped rows: {len(df_methaq) - len(data)}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Define all variations of missing/none/nan values
missing_indicators = [None, np.nan, 'None', 'none', 'nan', 'NaN', 'null', 'Null']

# 2. Count occurrences of these indicators across all columns
missing_counts = data.isin(missing_indicators).sum()
missing_percent = (missing_counts / len(data)) * 100

# 3. Create the missing values summary DataFrame
missing_df = pd.DataFrame({'Missing Counts': missing_counts, 'Percentage (%)': missing_percent})
print("--- Missing Values (including None and NaN strings) ---")
print(missing_df)

# 4. Plotting
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 5))
sns.barplot(
    x=missing_df.index, 
    y='Percentage (%)', 
    data=missing_df, 
    palette="viridis", 
    hue=missing_df.index, 
    legend=False
)
plt.title("Percentage of Missing/None Values per Column")
plt.xticks(rotation=45, ha='right')
plt.ylabel("Missing (%)")
plt.tight_layout()
plt.show()
```

```python
# 3. EDA - Distribution of top 10 Car Makes
make_counts = data['make'].value_counts()
top_10_makes = make_counts.head(10)
print("--- Top 10 Vehicle Makes ---")
for make, count in top_10_makes.items():
    pct = (count / len(data)) * 100
    print(f"{make}: {count} ({pct:.2f}%)")

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_makes.values, y=top_10_makes.index, palette="Blues_r", hue=top_10_makes.index, legend=False)
plt.title("Top 10 Vehicle Makes")
plt.xlabel("Count")
plt.ylabel("Make")
plt.show()
```

```python
# 4. EDA - Manufacturing Year Distribution
data['year'] = pd.to_numeric(data['year'], errors='coerce')
year_df = data.dropna(subset=['year'])
print("--- Year Statistics ---")
print(year_df['year'].describe())

plt.figure(figsize=(12, 5))
filtered_years = year_df[(year_df['year'] >= 1980) & (year_df['year'] <= 2026)]['year']
sns.histplot(filtered_years, bins=30, kde=True, color="#4c72b0", edgecolor="white")
plt.title("Distribution of Vehicle Manufacturing Years (1980 - 2026)")
plt.xlabel("Year")
plt.ylabel("Count")
plt.show()
```

```python
# 5. EDA - Distribution of top 10 Body Types
body_type_counts = data['bodyType'].value_counts()
top_10_body = body_type_counts.head(10)
print("--- Top 10 Body Types ---")
for bt, count in top_10_body.items():
    pct = (count / len(data)) * 100
    print(f"{bt}: {count} ({pct:.2f}%)")

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_body.values, y=top_10_body.index, palette="magma", hue=top_10_body.index, legend=False)
plt.title("Top 10 Vehicle Body Types")
plt.xlabel("Count")
plt.ylabel("Body Type")
plt.show()
```

```python
# 6. EDA - Engine Cylinders
data['cylinders'] = pd.to_numeric(data['cylinders'], errors='coerce')
cylinder_counts = data['cylinders'].value_counts().sort_index()
print("--- Cylinders Distribution ---")
print(cylinder_counts)

plt.figure(figsize=(10, 5))
sns.countplot(data=data, x='cylinders', palette="rocket", order=cylinder_counts.index, hue='cylinders', legend=False)
plt.title("Distribution of Engine Cylinders")
plt.xlabel("Cylinders")
plt.ylabel("Count")
plt.show()
```

```python
# 7. EDA - Regional Specifications
regional_counts = data['regionalSpec'].value_counts()
print("--- Regional Specs Distribution ---")
print(regional_counts)

plt.figure(figsize=(8, 8))
plt.pie(regional_counts.values, labels=regional_counts.index, autopct='%1.1f%%', 
        colors=sns.color_palette("pastel")[0:len(regional_counts)], startangle=140,
        wedgeprops={'edgecolor': 'white', 'linewidth': 1.5})
plt.title("Distribution of Regional Specifications")
plt.show()
```
