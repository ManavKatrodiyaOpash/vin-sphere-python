```python
CURRENT_YEAR = 2026

DEPR_RATES = {
    'nissan-patrol-safari': 0.10,
    'nissan-xterra': 0.11,
    'nissan-patrol-pickup': 0.11,
    'nissan-navara': 0.11,
    'nissan-urvan': 0.11,
    'nissan-gt-r': 0.13,
    'nissan-pickup': 0.12,
    'nissan-patrol': 0.13,
    'nissan-370z': 0.13,
    'nissan-tiida': 0.13,
    'nissan-x-trail': 0.13,
    'nissan-sunny': 0.14,
    'nissan-juke': 0.14,
    'nissan-kicks': 0.14,
    'nissan-sentra': 0.15,
    'nissan-altima': 0.15,
    'nissan-altima-coupe': 0.15,
    'nissan-pathfinder': 0.15,
    'nissan-maxima': 0.15,
    'nissan-z': 0.15,
    'nissan-patrol-pro-4x': 0.13,
    'nissan-patrol-nismo': 0.13,
    'nissan-magnite': 0.14,
}

df['car_age'] = CURRENT_YEAR - df['year']

def calc_depr(row):
    slug = row['model_slug']
    if slug not in DEPR_RATES:
        return np.nan
    rate = DEPR_RATES[slug]
    age = row['car_age']
    if age <= 0:
        return round(row['price_avg_aed'], 0)
    return round(row['price_avg_aed'] * ((1 - rate) ** age), 0)

df['depreciated_value'] = df.apply(calc_depr, axis=1)
```
