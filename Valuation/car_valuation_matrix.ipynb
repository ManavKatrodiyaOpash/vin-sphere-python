```python
"""
DriveArabia UAE Toyota Price Scraper
=====================================
Scrapes all Toyota models, their trims, and prices from:
  https://www.drivearabia.com/carprices/uae/toyota/{model}/{year}/

Output: data/drivearabia_toyota_prices.csv
Columns: model_slug, year, trim_name, price_aed, currency, url, scraped_at

Usage:
    python scrape_drivearabia_toyota.py
    python scrape_drivearabia_toyota.py --years 2024 2025 2026
    python scrape_drivearabia_toyota.py --models toyota-land-cruiser toyota-camry
    python scrape_drivearabia_toyota.py --delay 2.0 --output my_prices.csv
"""

import argparse
import csv
import logging
import re
import time
from datetime import datetime
from pathlib import Path

import requests
from bs4 import BeautifulSoup

# ─── Config ───────────────────────────────────────────────────────────────────

BASE_URL = "https://www.drivearabia.com"
TOYOTA_INDEX = f"{BASE_URL}/carprices/uae/toyota/"

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0.0.0 Safari/537.36"
    ),
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": BASE_URL,
}

DEFAULT_YEARS = [2022, 2023, 2024, 2025, 2026]
DEFAULT_DELAY = 1.5  # seconds between requests

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    datefmt="%H:%M:%S",
)
log = logging.getLogger(__name__)


# ─── Session ──────────────────────────────────────────────────────────────────

def make_session() -> requests.Session:
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session: requests.Session, url: str, retries: int = 3) -> BeautifulSoup | None:
    for attempt in range(1, retries + 1):
        try:
            resp = session.get(url, timeout=15)
            if resp.status_code == 200:
                return BeautifulSoup(resp.text, "lxml")
            elif resp.status_code == 404:
                log.debug("404 %s", url)
                return None
            elif resp.status_code == 429:
                wait = 10 * attempt
                log.warning("Rate-limited. Sleeping %ds", wait)
                time.sleep(wait)
            else:
                log.warning("HTTP %d for %s (attempt %d)", resp.status_code, url, attempt)
        except requests.RequestException as e:
            log.warning("Request error %s (attempt %d): %s", url, attempt, e)
            time.sleep(3 * attempt)
    return None


# ─── Discovery: all Toyota model slugs ───────────────────────────────────────

def discover_models(session: requests.Session) -> list[str]:
    """
    Parse the Toyota index page and collect all model slugs.
    Expected pattern: /carprices/uae/toyota/<model-slug>/
    """
    log.info("Discovering Toyota models from %s", TOYOTA_INDEX)
    soup = fetch(session, TOYOTA_INDEX)
    if not soup:
        log.error("Failed to fetch Toyota index page")
        return []

    slugs = set()
    pattern = re.compile(r"^/carprices/uae/toyota/(toyota-[^/]+)/$")

    for a in soup.find_all("a", href=True):
        m = pattern.match(a["href"])
        if m:
            slugs.add(m.group(1))

    # Fallback: try to find model links in any container (some sites
    # render model cards with data attributes or JS-injected markup)
    if not slugs:
        log.warning(
            "No model links found via <a> tags — page may be JS-rendered. "
            "Falling back to KNOWN_MODELS list."
        )
        return KNOWN_MODELS

    models = sorted(slugs)
    log.info("Found %d Toyota models", len(models))
    return models


# ─── Known models fallback (in case index page is JS-rendered) ───────────────

KNOWN_MODELS = [
    "toyota-4runner",
    "toyota-avalon",
    "toyota-avanza",
    "toyota-c-hr",
    "toyota-camry",
    "toyota-corolla",
    "toyota-corolla-cross",
    "toyota-fortuner",
    "toyota-granvia",
    "toyota-hiace",
    "toyota-hilux",
    "toyota-innova",
    "toyota-land-cruiser",
    "toyota-land-cruiser-70-series",
    "toyota-land-cruiser-prado",
    "toyota-prado",
    "toyota-rav4",
    "toyota-rush",
    "toyota-sequoia",
    "toyota-starlet",
    "toyota-tundra",
    "toyota-vios",
    "toyota-yaris",
    "toyota-yaris-cross",
]


# ─── Parsing: extract trims + prices from a model/year page ──────────────────

def parse_price_aed(raw: str) -> int | None:
    """
    Normalise price strings like:
        "AED 245,000"  →  245000
        "245,000 AED"  →  245000
        "From AED 99,900"  →  99900
    Returns None if no plausible price found (4–8 digit number).
    """
    # Extract first standalone number that looks like a vehicle price (4–8 digits)
    m = re.search(r"\b(\d{1,3}(?:,\d{3})*|\d{4,8})\b", raw)
    if not m:
        return None
    digits = re.sub(r"[^\d]", "", m.group(1))
    if not digits or not (4 <= len(digits) <= 8):
        return None
    return int(digits)


def parse_trims(soup: BeautifulSoup, model_slug: str, year: int, url: str) -> list[dict]:
    """
    Extract trim rows from a DriveArabia car price page.

    DriveArabia's typical structure (as of 2025):
    - A <table> or <div>-based list where each row has:
        - trim/variant name
        - price in AED
    We try several selectors in priority order.
    """
    rows = []
    now = datetime.utcnow().isoformat(timespec="seconds") + "Z"

    def make_row(trim_name: str, price_raw: str) -> dict:
        price_aed = parse_price_aed(price_raw)
        return {
            "model_slug": model_slug,
            "year": year,
            "trim_name": trim_name.strip(),
            "price_aed": price_aed,
            "price_raw": price_raw.strip(),
            "currency": "AED",
            "url": url,
            "scraped_at": now,
        }

    # ── Strategy 1: <table> rows ──────────────────────────────────────────────
    for table in soup.find_all("table"):
        for tr in table.find_all("tr"):
            tds = tr.find_all("td")
            if len(tds) >= 2:
                trim = tds[0].get_text(strip=True)
                price = tds[1].get_text(strip=True)
                if trim and ("AED" in price or re.search(r"\d{3,}", price)):
                    rows.append(make_row(trim, price))

    if rows:
        return rows

    # ── Strategy 2: repeated div pattern (variant + price pairs) ─────────────
    #   <div class="...variant...">Trim Name</div>
    #   <div class="...price...">AED 245,000</div>
    price_divs = soup.find_all(
        "div",
        class_=re.compile(r"price|variant|trim|spec|model[-_]row", re.I),
    )
    for div in price_divs:
        text = div.get_text(separator=" | ", strip=True)
        # Look for "SomeTrimName | AED 123,456" pattern
        parts = [p.strip() for p in text.split("|")]
        for i, part in enumerate(parts):
            if "AED" in part or re.search(r"\d{5,}", part):
                trim = parts[i - 1] if i > 0 else "Unknown"
                rows.append(make_row(trim, part))

    if rows:
        return rows

    # ── Strategy 3: generic text scan for price lines ─────────────────────────
    #   Last resort — look for short text nodes containing "AED \d+" only
    all_text_nodes = soup.find_all(
        string=re.compile(r"AED\s*[\d,]+", re.I)
    )
    for node in all_text_nodes:
        price_raw = node.strip()
        # Skip bloated nodes (e.g. full page text, script blobs)
        if len(price_raw) > 80:
            continue
        # Must contain a plausible price (AED + 4-8 digit number)
        if not re.search(r"AED\s*[\d,]{4,12}", price_raw, re.I):
            continue
        # Walk up to find a sibling/parent with trim name
        parent = node.parent
        trim = ""
        for sib in parent.find_previous_siblings(limit=3):
            candidate = sib.get_text(strip=True)
            if candidate and not re.search(r"AED|\d{5,}", candidate):
                trim = candidate
                break
        if not trim and parent.parent:
            for sib in parent.parent.find_previous_siblings(limit=2):
                candidate = sib.get_text(strip=True)
                if candidate and not re.search(r"AED|\d{5,}", candidate):
                    trim = candidate
                    break
        if price_raw:
            rows.append(make_row(trim or "Unknown", price_raw))

    return rows


# ─── Main scrape loop ─────────────────────────────────────────────────────────

def scrape(
    models: list[str],
    years: list[int],
    delay: float,
    output_path: Path,
) -> None:
    session = make_session()
    all_records = []
    total = len(models) * len(years)
    done = 0

    output_path.parent.mkdir(parents=True, exist_ok=True)

    for model in models:
        for year in years:
            url = f"{BASE_URL}/carprices/uae/toyota/{model}/{year}/"
            done += 1
            log.info("[%d/%d] %s", done, total, url)

            soup = fetch(session, url)
            if soup is None:
                log.debug("Skipping %s %s (no data)", model, year)
                time.sleep(delay * 0.5)
                continue

            records = parse_trims(soup, model_slug=model, year=year, url=url)
            if records:
                log.info("  → %d trims found", len(records))
                all_records.extend(records)
            else:
                log.warning("  → 0 trims parsed — page structure may differ. "
                            "Enable DEBUG logging and inspect the page HTML.")

            time.sleep(delay)

    if not all_records:
        log.error("No records scraped. Exiting without writing file.")
        return

    fieldnames = [
        "model_slug", "year", "trim_name", "price_aed",
        "price_raw", "currency", "url", "scraped_at",
    ]
    with output_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_records)

    log.info("Wrote %d records → %s", len(all_records), output_path)


# ─── CLI ──────────────────────────────────────────────────────────────────────

def main(
    years: list[int] | None = None,
    models: list[str] | None = None,
    delay: float = DEFAULT_DELAY,
    output: str = "data/drivearabia_toyota_prices.csv",
    debug: bool = False,
) -> None:
    """
    Can be called directly (Jupyter / import) or via CLI.

    Jupyter usage:
        from scrape_drivearabia_toyota import main
        main(years=[2025, 2026], delay=1.0)
    """
    if debug:
        logging.getLogger().setLevel(logging.DEBUG)

    resolved_years = years if years is not None else DEFAULT_YEARS

    session = make_session()

    if models:
        resolved_models = models
        log.info("Using %d user-specified models", len(resolved_models))
    else:
        resolved_models = discover_models(session)
        if not resolved_models:
            log.error("Model discovery failed and no models specified. Aborting.")
            return

    scrape(
        models=resolved_models,
        years=resolved_years,
        delay=delay,
        output_path=Path(output),
    )


def _cli_main() -> None:
    """Entry point for command-line use only. Skipped when run inside Jupyter."""
    import argparse

    parser = argparse.ArgumentParser(
        description="Scrape Toyota vehicle prices from DriveArabia UAE"
    )
    parser.add_argument(
        "--years", nargs="+", type=int, default=DEFAULT_YEARS,
        help="Model years to scrape (default: 2022-2026)",
    )
    parser.add_argument(
        "--models", nargs="+", default=None,
        help="Specific model slugs (default: auto-discover from index page)",
    )
    parser.add_argument(
        "--delay", type=float, default=DEFAULT_DELAY,
        help=f"Seconds between requests (default: {DEFAULT_DELAY})",
    )
    parser.add_argument(
        "--output", default="data/drivearabia_toyota_prices.csv",
        help="Output CSV path",
    )
    parser.add_argument(
        "--debug", action="store_true",
        help="Enable DEBUG logging",
    )
    # parse_known_args ignores Jupyter's --f=kernel.json injected arg
    args, _ = parser.parse_known_args()

    main(
        years=args.years,
        models=args.models,
        delay=args.delay,
        output=args.output,
        debug=args.debug,
    )


if __name__ == "__main__":
    _cli_main()
```

```python
"""
DriveArabia UAE Toyota Price Scraper
=====================================
Scrapes all Toyota models × all available years (2010 → current year).
For each model/year page, extracts every trim name + AED price.

Output CSV columns:
    model_slug | model_name | year | trim_name | price_aed | url | scraped_at

Usage (Jupyter):
    from scrape_drivearabia_toyota import main
    main()                                        # all models, 2010–current
    main(years=range(2022, 2027))                 # specific year range
    main(models=["toyota-land-cruiser"])          # one model, all years
    main(delay=1.0, output="data/prices.csv")

Usage (CLI):
    python scrape_drivearabia_toyota.py
    python scrape_drivearabia_toyota.py --start-year 2020
    python scrape_drivearabia_toyota.py --models toyota-camry toyota-yaris
    python scrape_drivearabia_toyota.py --delay 2.0 --output prices.csv
"""

import csv
import logging
import re
import time
from datetime import datetime
from pathlib import Path

import requests
from bs4 import BeautifulSoup

# ─── Constants ────────────────────────────────────────────────────────────────

BASE_URL       = "https://www.drivearabia.com"
TOYOTA_INDEX   = f"{BASE_URL}/carprices/uae/toyota/"
CURRENT_YEAR   = datetime.utcnow().year
DEFAULT_START  = 2010
DEFAULT_DELAY  = 1.5

HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/124.0.0.0 Safari/537.36"
    ),
    "Accept":          "text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer":         BASE_URL,
}

# ─── Strict regex patterns ────────────────────────────────────────────────────

# Model slug in index page href: /carprices/uae/toyota/toyota-<slug>/
RE_MODEL_HREF = re.compile(
    r"^/carprices/uae/toyota/(toyota-[a-z0-9]+(?:-[a-z0-9]+)*)/?$"
)

# AED price: optional prefix, then AED + digits/commas, 5–7 digit value
# Matches: "AED 245,000" / "AED245,000" / "From AED 99,900" / "245,000 AED"
RE_AED_PRICE = re.compile(
    r"(?:AED\s*(\d{1,3}(?:,\d{3})+|\d{5,7})"    # AED-first form
    r"|(\d{1,3}(?:,\d{3})+|\d{5,7})\s*AED)",     # digits-first form
    re.IGNORECASE,
)

# Trim name: 3–80 printable chars, not purely numeric, not a URL/email
RE_TRIM_NAME = re.compile(r"^[A-Za-z0-9][\w\s\-\/\.\+\(\)]{2,79}$")

# ─── Logging ──────────────────────────────────────────────────────────────────

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    datefmt="%H:%M:%S",
)
log = logging.getLogger(__name__)

# ─── Known models fallback ────────────────────────────────────────────────────

KNOWN_MODELS = [
    ("toyota-4runner",               "4Runner"),
    ("toyota-avalon",                "Avalon"),
    ("toyota-avanza",                "Avanza"),
    ("toyota-c-hr",                  "C-HR"),
    ("toyota-camry",                 "Camry"),
    ("toyota-corolla",               "Corolla"),
    ("toyota-corolla-cross",         "Corolla Cross"),
    ("toyota-fortuner",              "Fortuner"),
    ("toyota-granvia",               "Granvia"),
    ("toyota-hiace",                 "Hiace"),
    ("toyota-hilux",                 "Hilux"),
    ("toyota-innova",                "Innova"),
    ("toyota-land-cruiser",          "Land Cruiser"),
    ("toyota-land-cruiser-70-series","Land Cruiser 70 Series"),
    ("toyota-land-cruiser-prado",    "Land Cruiser Prado"),
    ("toyota-prado",                 "Prado"),
    ("toyota-rav4",                  "RAV4"),
    ("toyota-rush",                  "Rush"),
    ("toyota-sequoia",               "Sequoia"),
    ("toyota-starlet",               "Starlet"),
    ("toyota-tundra",                "Tundra"),
    ("toyota-vios",                  "Vios"),
    ("toyota-yaris",                 "Yaris"),
    ("toyota-yaris-cross",           "Yaris Cross"),
]

# ─── HTTP helpers ─────────────────────────────────────────────────────────────

def make_session() -> requests.Session:
    s = requests.Session()
    s.headers.update(HEADERS)
    return s


def fetch(session: requests.Session, url: str, retries: int = 3) -> BeautifulSoup | None:
    for attempt in range(1, retries + 1):
        try:
            resp = session.get(url, timeout=15)
            if resp.status_code == 200:
                return BeautifulSoup(resp.text, "lxml")
            if resp.status_code == 404:
                log.debug("404  %s", url)
                return None
            if resp.status_code == 429:
                wait = 15 * attempt
                log.warning("Rate-limited → sleeping %ds", wait)
                time.sleep(wait)
            else:
                log.warning("HTTP %d  %s  (attempt %d)", resp.status_code, url, attempt)
        except requests.RequestException as exc:
            log.warning("RequestError %s (attempt %d): %s", url, attempt, exc)
            time.sleep(4 * attempt)
    return None

# ─── Model discovery ──────────────────────────────────────────────────────────

def discover_models(session: requests.Session) -> list[tuple[str, str]]:
    """
    Returns list of (slug, display_name) tuples.
    Falls back to KNOWN_MODELS if index page is JS-rendered.
    """
    log.info("Discovering Toyota models → %s", TOYOTA_INDEX)
    soup = fetch(session, TOYOTA_INDEX)
    if not soup:
        log.warning("Index fetch failed — using KNOWN_MODELS fallback")
        return KNOWN_MODELS

    seen: dict[str, str] = {}
    for a in soup.find_all("a", href=True):
        href = a["href"].rstrip("/")
        # Normalise to path-only
        if href.startswith("http"):
            href = "/" + "/".join(href.split("/")[3:])
        m = RE_MODEL_HREF.match(href + "/")
        if m:
            slug = m.group(1)
            label = a.get_text(strip=True) or slug.replace("toyota-", "").replace("-", " ").title()
            seen[slug] = label

    if not seen:
        log.warning("No model links found — using KNOWN_MODELS fallback")
        return KNOWN_MODELS

    result = sorted(seen.items())
    log.info("Found %d Toyota models", len(result))
    return result

# ─── Price extraction ─────────────────────────────────────────────────────────

def extract_price_aed(text: str) -> int | None:
    """
    Strict extraction: must match AED + 5–7 digit number (50k–9.9M range).
    Returns integer AED value or None.
    """
    m = RE_AED_PRICE.search(text)
    if not m:
        return None
    raw = (m.group(1) or m.group(2)).replace(",", "")
    if not raw.isdigit():
        return None
    val = int(raw)
    # Sanity range: AED 10,000 – AED 9,999,999
    if not (10_000 <= val <= 9_999_999):
        return None
    return val


def is_valid_trim(text: str) -> bool:
    """
    Reject nav labels, headers, and other non-trim strings.
    """
    if not text or not RE_TRIM_NAME.match(text):
        return False
    # Reject pure navigation / UI noise
    BLOCKLIST = {
        "home", "cars", "toyota", "price", "prices", "car prices",
        "new cars", "used cars", "compare", "reviews", "news",
        "specs", "specifications", "variants", "variant", "trim",
        "model", "models", "engine", "transmission", "fuel",
        "read more", "view all", "see all", "show more",
    }
    if text.lower() in BLOCKLIST:
        return False
    # Must contain at least one letter (not purely numeric)
    if not re.search(r"[A-Za-z]", text):
        return False
    return True


def parse_trims(soup: BeautifulSoup, model_slug: str, model_name: str,
                year: int, url: str) -> list[dict]:
    """
    Three-strategy cascade, each producing (trim_name, price_aed) pairs.
    Only rows passing both is_valid_trim() and extract_price_aed() are kept.
    Deduplication by (trim_name_lower, price_aed) within the page.
    """
    now = datetime.utcnow().isoformat(timespec="seconds") + "Z"
    seen: set[tuple[str, int | None]] = set()
    rows: list[dict] = []

    def emit(trim_name: str, price_text: str) -> None:
        trim_name = trim_name.strip()
        if not is_valid_trim(trim_name):
            return
        price_aed = extract_price_aed(price_text)
        key = (trim_name.lower(), price_aed)
        if key in seen:
            return
        seen.add(key)
        rows.append({
            "model_slug":  model_slug,
            "model_name":  model_name,
            "year":        year,
            "trim_name":   trim_name,
            "price_aed":   price_aed,
            "url":         url,
            "scraped_at":  now,
        })

    # ── Strategy 1: <table> with ≥2 <td> columns ─────────────────────────────
    for table in soup.find_all("table"):
        for tr in table.find_all("tr"):
            tds = [td.get_text(strip=True) for td in tr.find_all("td")]
            if len(tds) < 2:
                continue
            trim  = tds[0]
            price = " ".join(tds[1:])   # join remaining cells in case price is split
            if RE_AED_PRICE.search(price):
                emit(trim, price)

    if rows:
        log.debug("  strategy=table  rows=%d", len(rows))
        return rows

    # ── Strategy 2: <li> / <div> class-hinted containers ─────────────────────
    PRICE_CLASS = re.compile(
        r"\b(price|variant|trim|spec|grade|model[-_]?row|car[-_]?item)\b", re.I
    )
    for container in soup.find_all(["li", "div"], class_=PRICE_CLASS):
        text = container.get_text(separator="|||", strip=True)
        parts = [p.strip() for p in text.split("|||") if p.strip()]
        # Find first part with a price, treat preceding part as trim
        for i, part in enumerate(parts):
            if RE_AED_PRICE.search(part):
                trim = parts[i - 1] if i > 0 else ""
                emit(trim, part)
                break   # one price per container

    if rows:
        log.debug("  strategy=div-class  rows=%d", len(rows))
        return rows

    # ── Strategy 3: text-node scan — strict short-node AED match only ─────────
    for node in soup.find_all(string=RE_AED_PRICE):
        price_raw = node.strip()
        # Hard length cap — skip anything that looks like a script/style blob
        if len(price_raw) > 60:
            continue
        # Must match strict AED pattern
        if not RE_AED_PRICE.search(price_raw):
            continue
        # Walk DOM for a nearby trim label
        trim = ""
        parent = node.parent
        # 1. prev sibling of parent
        for sib in (parent.find_previous_siblings(limit=3) if parent else []):
            candidate = sib.get_text(strip=True)
            if is_valid_trim(candidate) and not RE_AED_PRICE.search(candidate):
                trim = candidate
                break
        # 2. prev sibling of grandparent
        if not trim and parent and parent.parent:
            for sib in parent.parent.find_previous_siblings(limit=2):
                candidate = sib.get_text(strip=True)
                if is_valid_trim(candidate) and not RE_AED_PRICE.search(candidate):
                    trim = candidate
                    break
        emit(trim or "Unknown", price_raw)

    if rows:
        log.debug("  strategy=text-node  rows=%d", len(rows))

    return rows

# ─── Main scrape loop ─────────────────────────────────────────────────────────

def scrape(
    models:      list[tuple[str, str]],
    years:       list[int],
    delay:       float,
    output_path: Path,
) -> int:
    """Returns number of records written."""
    session    = make_session()
    all_records: list[dict] = []
    total = len(models) * len(years)
    n = 0

    output_path.parent.mkdir(parents=True, exist_ok=True)

    for slug, name in models:
        for year in years:
            n += 1
            url = f"{BASE_URL}/carprices/uae/toyota/{slug}/{year}/"
            log.info("[%d/%d] %s %d", n, total, slug, year)

            soup = fetch(session, url)
            if soup is None:
                time.sleep(delay * 0.4)
                continue

            records = parse_trims(soup, slug, name, year, url)
            if records:
                log.info("  → %d trims", len(records))
                all_records.extend(records)
            else:
                log.debug("  → 0 trims  (no price table found)")

            time.sleep(delay)

    if not all_records:
        log.error("No records scraped.")
        return 0

    fieldnames = ["model_slug", "model_name", "year", "trim_name",
                  "price_aed", "url", "scraped_at"]
    with output_path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=fieldnames)
        writer.writeheader()
        writer.writerows(all_records)

    log.info("Done — %d records → %s", len(all_records), output_path)
    return len(all_records)

# ─── Public entry point ───────────────────────────────────────────────────────

def main(
    years:      list[int] | range | None = None,
    models:     list[str] | None         = None,
    start_year: int                       = DEFAULT_START,
    delay:      float                     = DEFAULT_DELAY,
    output:     str                       = "data/drivearabia_toyota_prices.csv",
    debug:      bool                      = False,
) -> int:
    """
    Jupyter-friendly entry point. Returns record count.

    Args:
        years:      explicit list/range of years; overrides start_year
        models:     explicit slug list e.g. ["toyota-camry"]; None = auto-discover
        start_year: first year to scrape when years is None (default 2010)
        delay:      seconds between HTTP requests
        output:     CSV output path
        debug:      enable DEBUG logging
    """
    if debug:
        logging.getLogger().setLevel(logging.DEBUG)

    resolved_years = list(years) if years is not None else list(range(start_year, CURRENT_YEAR + 1))
    log.info("Year range: %d – %d  (%d years)", resolved_years[0], resolved_years[-1], len(resolved_years))

    session = make_session()

    if models:
        # Accept bare slugs; build (slug, display_name) tuples
        resolved_models = [
            (s if s.startswith("toyota-") else f"toyota-{s}",
             s.replace("toyota-", "").replace("-", " ").title())
            for s in models
        ]
        log.info("Using %d user-specified models", len(resolved_models))
    else:
        resolved_models = discover_models(session)
        if not resolved_models:
            log.error("Model discovery failed. Aborting.")
            return 0

    return scrape(
        models=resolved_models,
        years=resolved_years,
        delay=delay,
        output_path=Path(output),
    )

# ─── CLI ──────────────────────────────────────────────────────────────────────

def _cli_main() -> None:
    import argparse
    p = argparse.ArgumentParser(description="Scrape Toyota prices from DriveArabia UAE")
    p.add_argument("--start-year", type=int, default=DEFAULT_START,
                   help=f"First year to scrape (default {DEFAULT_START})")
    p.add_argument("--end-year",   type=int, default=CURRENT_YEAR,
                   help=f"Last year to scrape (default {CURRENT_YEAR})")
    p.add_argument("--years",  nargs="+", type=int, default=None,
                   help="Explicit year list (overrides --start/end-year)")
    p.add_argument("--models", nargs="+", default=None,
                   help="Specific model slugs (default: auto-discover)")
    p.add_argument("--delay",  type=float, default=DEFAULT_DELAY,
                   help=f"Seconds between requests (default {DEFAULT_DELAY})")
    p.add_argument("--output", default="data/drivearabia_toyota_prices.csv")
    p.add_argument("--debug",  action="store_true")
    args, _ = p.parse_known_args()   # parse_known_args → ignores Jupyter's --f= arg

    years = args.years or list(range(args.start_year, args.end_year + 1))
    main(years=years, models=args.models, delay=args.delay,
         output=args.output, debug=args.debug)


if __name__ == "__main__":
    _cli_main()
```
