```python
import pandas as pd
import numpy as np
```

```python
file_path = "../Data/df__methaq.csv"
data = pd.read_csv(file_path)

data.head()
```

```python
data = data.loc[data['make'].map(data['make'].value_counts()).sort_values(ascending=False).index]
data
```

```python
import pandas as pd

# Load dataset
# Create lookup dataframe
lookup_data = (
    data[
        ['chassisNumber','make', 'model', 'year', 'trim', 'price']
    ]
    .drop_duplicates(
        subset=['make', 'model', 'year', 'trim'],
        keep='first'
    )
    .reset_index(drop=True)
)

# View lookup table
print(lookup_data.head())

print("\nLookup Table Shape:")
print(lookup_data.shape)

# Create dictionary for fast valuation lookup
lookup_dict = lookup_data.set_index(
    ['make', 'model', 'year', 'trim']
)['price'].to_dict()

# Example lookup
make = "TOYOTA"
model = "CAMRY"
year = 2024
trim = "XLE"

price = lookup_dict.get(
    (make, model, year, trim)
)

print(f"\nPrice: {price}")
```
