```python
TOYOTA_DEPRECIATION_RATES = {

    # Economy Cars
    "toyota-yaris": 0.09,
    "toyota-yaris-sedan": 0.09,
    "toyota-raize": 0.09,
    "toyota-avanza": 0.10,
    "toyota-rush": 0.09,
    "toyota-veloz": 0.09,
    "toyota-liteace": 0.10,

    # Compact Sedans / Hatchbacks
    "toyota-corolla": 0.07,
    "toyota-corolla-cross": 0.07,
    "toyota-c-hr": 0.08,
    "toyota-urban-cruiser": 0.08,

    # Mid-Size Sedans
    "toyota-camry": 0.08,
    "toyota-aurion": 0.09,
    "toyota-avalon": 0.08,
    "toyota-crown": 0.07,

    # MPV / Van
    "toyota-innova": 0.07,
    "toyota-granvia": 0.08,
    "toyota-previa": 0.08,
    "toyota-hiace": 0.06,

    # SUVs
    "toyota-rav-4": 0.06,
    "toyota-fortuner": 0.05,
    "toyota-highlander": 0.06,
    "toyota-sequoia": 0.06,

    # Pickup Trucks
    "toyota-hilux": 0.05,
    "toyota-land-cruiser-pickup": 0.04,

    # Land Cruiser Family
    "toyota-land-cruiser": 0.04,
    "toyota-land-cruiser-gr-sport": 0.04,
    "toyota-land-cruiser-70": 0.03,
    "toyota-land-cruiser-prado": 0.05,
    "toyota-land-cruiser-prado-swb": 0.05,

    # Off-Road
    "toyota-fj-cruiser": 0.05,

    # Sports Cars
    "toyota-86": 0.08,
    "toyota-gr86": 0.08,
    "toyota-supra": 0.07,
    "toyota-gr-corolla": 0.07,
    "toyota-gr-yaris": 0.06,
}
```

```python
df["depreciation_rate"] = (
    df["model_slug"]
    .map(TOYOTA_DEPRECIATION_RATES)
    .fillna(0.08)
)

df["estimated_value"] = (
    df["price_avg_aed"]
    * ((1 - df["depreciation_rate"]) ** df["car_age"])
)
```

```python
def depreciation_factor(age, rate):
    if age <= 3:
        return (1 - rate * 1.3) ** age
    else:
        first_3 = (1 - rate * 1.3) ** 3
        remaining = (1 - rate) ** (age - 3)
        return first_3 * remaining
```

```python
df["estimated_value"] = df.apply(
    lambda x: x["price_avg_aed"] *
    depreciation_factor(
        x["car_age"],
        x["depreciation_rate"]
    ),
    axis=1
)
```

```python
CURRENT_YEAR = 2026

DEPR_RATES = {
    'toyota-86': 0.105,
    'toyota-aurion': 0.10,
    'toyota-avalon': 0.10,
    'toyota-avanza': 0.09,
    'toyota-c-hr': 0.09,
    'toyota-camry': 0.105,
    'toyota-corolla': 0.10,
    'toyota-corolla-cross': 0.115,
    'toyota-crown': 0.13,
    'toyota-fj-cruiser': 0.075,
    'toyota-fortuner': 0.09,
    'toyota-gr-corolla': 0.10,
    'toyota-gr-yaris': 0.095,
    'toyota-gr86': 0.13,
    'toyota-granvia': 0.09,
    'toyota-hiace': 0.095,
    'toyota-highlander': 0.105,
    'toyota-hilux': 0.075,
    'toyota-innova': 0.10,
    'toyota-land-cruiser': 0.10,
    'toyota-land-cruiser-70': 0.08,
    'toyota-land-cruiser-gr-sport': 0.095,
    'toyota-land-cruiser-pickup': 0.09,
    'toyota-land-cruiser-prado': 0.10,
    'toyota-land-cruiser-prado-swb': 0.10,
    'toyota-liteace': 0.095,
    'toyota-previa': 0.125,
    'toyota-raize': 0.115,
    'toyota-rav-4': 0.105,
    'toyota-rush': 0.10,
    'toyota-sequoia': 0.105,
    'toyota-supra': 0.08,
    'toyota-urban-cruiser': 0.125,
    'toyota-veloz': 0.115,
    'toyota-yaris': 0.095,
    'toyota-yaris-sedan': 0.095,
}

df['car_age'] = CURRENT_YEAR - df['year']

def calc_depr(row):
    slug = row['model_slug']
    if slug not in DEPR_RATES:
        return np.nan
    rate = DEPR_RATES[slug]
    age = row['car_age']
    if age <= 0:
        return round(row['price_avg_aed'], 0)
    return round(row['price_avg_aed'] * ((1 - rate) ** age), 0)

df['depreciated_value'] = df.apply(calc_depr, axis=1)
```
