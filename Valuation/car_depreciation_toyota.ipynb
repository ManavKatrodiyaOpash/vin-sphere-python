```python
import pandas as pd
import numpy as np

# ============================================================
# CONFIG
# ============================================================

CURRENT_YEAR = 2026

# Load Toyota dataset
df = pd.read_csv("data/drivearabia_toyota_prices_newest_final.csv")

# Load depreciation rates dataset
dep_df = pd.read_csv("data/annual_dep_rate.csv")

# ============================================================
# CLEAN DATA
# ============================================================

# Clean depreciation file
dep_df["make"] = dep_df["make"].astype(str).str.strip().str.upper()
dep_df["model"] = dep_df["model"].astype(str).str.strip().str.upper()

# ============================================================
# MODEL MAPPING
# ============================================================

MODEL_MAPPING = {
    "toyota-86": "86",
    "toyota-aurion": "AURION",
    "toyota-avalon": "AVALON",
    "toyota-avanza": "AVANZA",
    "toyota-c-hr": "C-HR",
    "toyota-camry": "CAMRY",
    "toyota-corolla": "COROLLA",
    "toyota-corolla-cross": "COROLLA CROSS",
    "toyota-crown": "CROWN",
    "toyota-fj-cruiser": "FJ CRUISER",
    "toyota-fortuner": "FORTUNER",
    "toyota-gr-corolla": "GR COROLLA",
    "toyota-gr-yaris": "GR YARIS",
    "toyota-gr86": "GR86",
    "toyota-granvia": "GRANVIA",
    "toyota-hiace": "HIACE",
    "toyota-highlander": "HIGHLANDER",
    "toyota-hilux": "HILUX",
    "toyota-innova": "INNOVA",
    "toyota-land-cruiser": "LAND CRUISER",
    "toyota-land-cruiser-70": "LAND CRUISER 70",
    "toyota-land-cruiser-gr-sport": "LAND CRUISER GR SPORT",
    "toyota-land-cruiser-pickup": "LAND CRUISER PICKUP",
    "toyota-land-cruiser-prado": "LAND CRUISER PRADO",
    "toyota-land-cruiser-prado-swb": "LAND CRUISER PRADO",
    "toyota-liteace": "LITEACE",
    "toyota-previa": "PREVIA",
    "toyota-raize": "RAIZE",
    "toyota-rav-4": "RAV4",
    "toyota-rush": "RUSH",
    "toyota-sequoia": "SEQUOIA",
    "toyota-supra": "SUPRA",
    "toyota-urban-cruiser": "URBAN CRUISER",
    "toyota-veloz": "VELOZ",
    "toyota-yaris": "YARIS",
    "toyota-yaris-sedan": "YARIS"
}

# ============================================================
# PREPARE TOYOTA DATA
# ============================================================

df["make"] = "TOYOTA"

df["model_name"] = (
    df["model_slug"]
    .map(MODEL_MAPPING)
    .astype(str)
    .str.upper()
)

# ============================================================
# CREATE DEPRECIATION LOOKUP
# ============================================================

dep_lookup = (
    dep_df.groupby(["make", "model"])["annual_dep_rate"]
    .mean()
    .reset_index()
)

# ============================================================
# MERGE DEPRECIATION RATES
# ============================================================

df = df.merge(
    dep_lookup,
    left_on=["make", "model_name"],
    right_on=["make", "model"],
    how="left"
)

# ============================================================
# CALCULATE CAR AGE
# ============================================================

df["car_age"] = CURRENT_YEAR - df["year"]

# Prevent negative ages
df["car_age"] = df["car_age"].clip(lower=0)

# ============================================================
# CALCULATE DEPRECIATED VALUE
# ============================================================

def calc_depreciated_value(row):

    rate = row["annual_dep_rate"]

    if pd.isna(rate):
        return np.nan

    age = row["car_age"]

    if age == 0:
        return round(row["price_avg_aed"], 0)

    depreciated_value = (
        row["price_avg_aed"] *
        ((1 - rate) ** age)
    )

    return round(depreciated_value, 0)

df["depreciated_value"] = df.apply(
    calc_depreciated_value,
    axis=1
)

# ============================================================
# OPTIONAL CHECKS
# ============================================================

print("Total Rows:", len(df))
print("Matched Depreciation Rates:", df["annual_dep_rate"].notna().sum())
print("Missing Depreciation Rates:", df["annual_dep_rate"].isna().sum())

# Show unmatched models
missing_models = (
    df[df["annual_dep_rate"].isna()]
    ["model_slug"]
    .drop_duplicates()
    .tolist()
)

print("\nUnmatched Models:")
print(missing_models)

# ============================================================
# SAVE OUTPUT
# ============================================================

df.to_csv(
    "data/drivearabia_toyota_prices_with_depreciation_aanual_rate.csv",
    index=False
)

print("\nDone!")
print("Saved as: drivearabia_toyota_prices_with_depreciation.csv")
```
