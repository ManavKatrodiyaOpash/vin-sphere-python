```python
import pandas as pd
```

```python
data = pd.read_csv("../Data/df_.csv")
data
```

```python
data = data[["chassisNumber", "make", "model", "trim", "bodyType", "cylinders", "year", "origin","weightInKg", "color", "noOfPassengers", "price"]]
```

```python
data
```

```python


for col in data.columns:
    if data[col].dtype == "object":
        data[col] = [x.upper() if isinstance(x, str) else x for x in data[col]]

print(data.head())
```

```python
data
```

```python
data = data.loc[data['make'].map(data['make'].value_counts()).sort_values(ascending=False).index]
```

```python
data.drop_duplicates(subset=['chassisNumber'], keep='first', inplace=True)
```

```python
data["chassis_len"] = data["chassisNumber"].str.len()

is_standard = (data["chassis_len"] == 17) 
data = data[is_standard]
data
```
