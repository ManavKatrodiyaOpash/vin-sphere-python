```python
import pandas as pd

data = pd.read_csv(r"../data/data_methaq(1.0).csv")

for col in data.columns:
    if data[col].dtype == "object":
        data[col] = [x.upper() if isinstance(x, str) else x for x in data[col]]

print(data.head())
```

# checking chassis number length and 10th pos

```python
data["chassis_len"] = data["chassisNumber"].str.len()
data["pos10"] = data["chassisNumber"].str[9]

is_standard = (data["chassis_len"] == 17) & (~data["pos10"].isin(['I', 'O', 'Q', 'U', 'Z', '0']))
data = data[is_standard]
data
```

```python
from IPython.display import display

for pos_4 in data['pos4'].dropna().unique():

    # 1. Filter the DataFrame for the current pos4 value
    filtered_df = data.loc[
        data['pos4'] == pos_4,
        ['make', 'model', 'cylinders', 'bodyType', 'pos4', 'trim', 'WMI', 'regionalSpec', 'noOfDoors', 'VDS']
    ].reset_index(drop=True)

    # 2. Header for the current pos4 group
    print("\n" + "=" * 100)
    print(f"POS4: {pos_4}")
    print("=" * 100)

    # 3. Print unique values for all object columns (except pos4)
    print("--- Unique Values for Object Columns ---")
    object_cols = filtered_df.select_dtypes(include=['object']).columns
    
    for col in object_cols:
        if col != ['pos4', 'VDS']:
            unique_vals = filtered_df[col].dropna().unique()
            print(f"  {col}: {list(unique_vals)}")
            
    print("-" * 100)

    # 4. Display the filtered DataFrame
    display(filtered_df)
```
